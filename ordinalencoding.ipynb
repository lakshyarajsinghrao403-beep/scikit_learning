ORDINAL ENCODING ON SIZE

```python
sizes = ['Small', 'Medium', 'Large']
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
enc = OrdinalEncoder(categories= [sizes])
```

```python
enc.fit_transform(df[['size']])     #it is gonna make an array but we need it in our dataset
```

```python
df.head()   #nothing has changed
```

```python
df['size'] = enc.fit_transform(df[['size']])    #that's how it's implemented in the dataset
```
