```python
import pandas as pd
```

```python
d = {'sales': [100000,222000,1000000,522000,111111,222222,1111111,20000,75000,90000,1000000,10000],
'city': ['Tampa','Tampa','Orlando','Jacksonville','Miami','Jacksonville','Miami','Miami','Orlando','Orlando','Orlando','Orlando'],
'size': ['Small', 'Medium','Large','Large','Small','Medium','Large','Small','Medium','Medium','Medium','Small',]}
```

```python
df = pd.DataFrame(data = d)
df.head()
```

```python
df['city'].nunique()
```

gonna use one hot encoding on city

```python
df['city'].value_counts()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(handle_unknown= 'ignore',sparse_output= False).set_output(transform='pandas')

#handle_unknown =if i get a new column in testing which i didn't work on training , don't give me an error
#sparse_output = sparse is way of storing large data where it stores only the null values (good for large data)
#but we have small data so i wouldn't sparse my data
#with set_output i would get transformed data as pandas dataframe, without it i would get a numpy array
```

```python
ohetransform = ohe.fit_transform(df[['city']])   #always two dimensional : df[['city']]
```

```python
ohetransform
```

```python
df = pd.concat([df,ohetransform],axis=1).drop(columns = ['city']) 
```
