```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_digits
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
digits = load_digits()
```

```python
dir(digits)
```

```python
digits.images[0]     #it's a picture but it will show a array of 8 * 8
```

```python
#to get the picture
for i in range(3):
    plt.matshow(digits.images[i])   #matshow() lets us see the image
```

```python
digits.target[0:3]    #images is the feature and target is the target(which is to be predicted)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(digits.data, digits.target,test_size= 0.2,random_state=11)
```

IN MULTICLASS CLASSIFICATION , PREDICTION IS GONNA BE OUT OF MULTIPLE THINGS

```python
from sklearn.linear_model import LogisticRegression
```

```python
model  = LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_test,y_test)
```
